## 인과추론의 개념

**인과추론** : 연관관계로부터 인과를 추론하고 "언제", "왜" 다른지 이해하는 과학

- 연관관계 != 인관관계
- 연관관계 ⊃ 인과관계 (일부 예외 케이스 있음)

머신러닝은 인과추론으로 답을 얻으려는 '만약(What if)' 유형의 질문에 취약하지만, 예측 문제를 다룰 때는 매우 유용함

## 연관관계와 인과관계

인과추론의 근본적인 문제 : 동일한 실험 대상이 **처치를 받은 상태와 받지 않은 상태를 동시에 관측할 수 없다**는 것

- 처치 (treatment) : 구하려는 효과에 대한 개입을 나타낼 때 사용하는 용어
- 반사실(Counterfactual): 실제로 일어난 일(T=1)과 정반대의 선택(T=0)을 했을 때 나타났을 잠재적 결과. 현실에서 직접 관찰할 수 없기 때문에, 인과추론의 모든 기법은 결국 "관찰 불가능한 반사실을 가장 정확하게 추정하는 방법"이다. (feat. gemini)

**인과모델(causal model)** : 화살표($\leftarrow$)로 표시하는 일련의 할당 메커니즘으로, **화살표를 사용하여 인과관계의 비가역성(nonreversibility)을 분명하게 표시**

- $T \leftarrow f_t(u_t)$
- $Y \leftarrow f_y(T,u_y)$

T는 외부 요인 $u_t$로 생성되고, 결과 $Y$는 $T$와 또 다른 외부 요인 $u_y$를 재료삼아 생성됨

식으로 표현하면 $Y = a + b_1T_i + e_i$

- 작성한 식은 선형모델(linear model)
- 선형모델 : 인과 결과 같은 변수들 사이의 관계에 선형 함수를 부여해 단순화한 모델

**개입(Intervention)** : 인과적 효과를 얻기 위해 원인($T$)을 특정 값으로 강제 고본해 보는 조건 설정($do(T=t_0)$)
- $T \leftarrow t_0$
- $Y \leftarrow f_y(t_0, u_y)$
- $do(.)$ 연산자 : 관측된 데이터에서 항상 얻을 수 없는 인과 추정량(causal quantity)을 정의하는 데 사용

**식별(Identification)** : 인과 추정량에 대한 이론적 표현에서 직접 관측할 수 없는 부분(반사실 등)을 제거하여, 관측 가능한 데이터만으로 계산할 수 있게 만드는 과정
